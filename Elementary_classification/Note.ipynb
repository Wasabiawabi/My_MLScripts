# 人工ニューロン(パーセプトロン)
パーセプトロンは、脳の神経細胞を模したモデル。
### パーセプトロンの使用用途
入力に基づいた出力が0か1になる**二値分類**に用いることができる(出力が0や1の離散値の場合は**分類問題**、出力が連続値の場合は**回帰問題**)。<br>
二値分類とは、**入力をクラス0とクラス1の2クラスに分類**すること。<br>
使い物になるかどうかまで考慮すると、1本の直線や平らな面でデータに対する境界線を求められる**線形分類問題**に適用できる。<br>
逆に、1本の直線や平らな面では境界線が決められないような**非線形分類問題**には適応できない。<br>
### パーセプトロンの構造
入力を1つ与えると、出力が1つ得られる。<br>
出力の種類は、0か1の2種類あり、入力の信号(離連続値)がニューロンに与えられると、0か1(離散値)が出力される。<br>
ニューロンには閾値が存在し、入力値が閾値以上ならば出力は1、閾値未満なら出力は0となる。<br>
入力に対する出力に正解がある前提で、出力が正解とことなるパターンを無くすことを目指す。
### パーセプトロンの原理解説
##### データ構造
**入力値ベクトル**$x$が行列として入力されると考える。<br>
**入力値**は入力値ベクトルの各1行にあたり、神経細胞に与えられる外部からの1刺激と考えられる。<br>
1刺激は任意の個数(列数)の値が合わさったものと考えられる。<br>
入力値の列は**特徴量**と呼ばれ、1つ(1行)の入力値がn列の場合、n個の特徴量があるということになり、**n次元**のデータとなる。<br>
パーセプトロンの学習では、**重みベクトル**にある各**重み**(各特徴量に対する感度)は最適値を求めたい**パラメータ**となる。

$$
\begin{gather}
入力値ベクトル(m行n列)\ \ x= 
\left[
\begin{array}{}
  x_{11}&x_{12}&\cdots&x_{1n}\\
  x_{21}&x_{22}&\cdots&x_{2n}\\
  \vdots&\vdots&\ddots&\vdots\\
  x_{m1}&x_{m2}&\cdots&x_{mn}
\end{array}
\right] \\ \ \
重みベクトル(1行n列)\ \ w= \ \ \left[
\begin{array}{}
w_1 & \ w_2 \ &\cdots & w_n
\end{array}
\right]
\end{gather}
$$
##### 入力と出力
1行の入力値と重みベクトルのドット積(内積)を**総入力**$z$として、これを最終的な入力とする。
$$
m行目の総入力(ドット積)\ \ z_m= w\cdot x_m = w_1x_{m1} + w_2x_{m2} + \cdots + w_nx_{mn}
$$


**閾値関数**$σ(z)$は、総入力$z$と**閾値**$θ$を基に0か1を出力する。<br>
閾値は、神経細胞の発火しやすさを表していて、出力0は発火していない状態、出力1は発火した状態と言い換えられる。<br>
パーセプトロンの学習では、閾値関数の閾値が、重みベクトルと同様、最適値を求めたいパラメータとなる。
$$
決定関数 \ \ σ(z_m) =
\begin{cases}
1,\ \ (z_m \ge θ) \\
0,\ \ (z_m \lt θ)
\end{cases}
$$
##### 式変形する
重みベクトルにある各重み(各特徴量に対する感度)と閾値を調整することで、出力(神経細胞の反応)が得られるようにする。<br>
入力データから得られた出力(0か1)に応じて、パラメータである重みベクトル$w$と閾値$θ$を更新する。<br>
**バイアスユニット**を$b=-θ$として、閾値を総入力に組み込むことで、全てのパラメータが1つの式で更新できるようになる。<br>
バイアスユニットは総入力全体の調整役と捉えることができる。
$$
\begin{gather}
閾値の式変形(b=-θ)\\
z \ge θ \iff z-θ \ge0 \iff z+b\ge0\\
z < θ \iff z-θ<0 \iff z+b<0\\\\
総入力zを再定義\\
z_m = w\cdot x_m + b\\\\
決定関数はこうなる\\
 \ \ σ(z_m) =
\begin{cases}
1,\ \ (z_m \ge 0) \\
0,\ \ (z_m \lt 0)
\end{cases}
\end{gather}
$$

### パーセプトロンの実装
##### データ構造
$i$行$j$列の入力データと、行ごとの正解データが与えられるとする。<br>
重みベクトル$w$とバイアス$b$を学習で更新したい。
$$

\begin{gather}
事前に与えられるデータ \\
入力データ\ 
x= 
\left[
\begin{array}{}
  x_{11}&x_{12}&\cdots&x_{1j}\\
  x_{21}&x_{22}&\cdots&x_{2j}\\
  \vdots&\vdots&\ddots&\vdots\\
  x_{i1}&x_{i2}&\cdots&x_{ij}
\end{array}
\right],\ \
正解データ\ y=\left[
\begin{array}{}
y_1\\
y_2\\
\vdots\\
y_i
\end{array}
\right]\\\\
学習で求めたい値\\
重みベクトル\ \ w=\left[
\begin{array}{}
w_1 & w_2 & \cdots & w_j
\end{array}
\right],\ \ 
バイアス\ b\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \
\end{gather}
$$
##### 学習則
各入力データから得られる出力と正解データができる限り一致するように、重みベクトル$w$とバイアス$b$を更新することを考える。<br>
ここで、$\hat y_i$を入力$x_i$と総入力$z$から得られた**予測値**(出力)とすると、**予測値ベクトル**は$\hat y$となり、正解データ$y$と同じ形状になる。<br>
予測値は、パラメータ更新前の重みベクトル$w$、バイアス$b$と入力データから得られる総入力$z$を、閾値関数$σ$に代入することで得られる。<br>
$\alpha$は**学習率=learning rate**で、更新値をどれだけ反映させるかを決める(プログラムでは、$\alpha$以外に$eta$と書かれることもある)。
$$
\begin{gather}
予測値ベクトル\\
\hat y=\left[
\begin{array}{}
\hat y_1\\
\hat y_2\\
\vdots\\
\hat y_i
\end{array}
\right]
=\left[
\begin{array}{}
σ(z_1)\\
σ(z_2)\\
\vdots\\
σ(z_i)
\end{array}
\right]
\\ \\
総入力(i行j列版) \\
z_i=w \cdot x_i+b \\ \\
重みベクトルとバイアスの更新式\\
w\leftarrow w+\Delta w\\
b\leftarrow b+\Delta b\ \\\\
更新値の計算式(\alpha は任意の値)\\
error=(y-\hat y)\\
\Delta w=\alpha(error^T\cdot x_{i})\\
\Delta b=\alpha \sum_i error_i
\end{gather}
$$
今回は0か1が出力のため、予測値$\hat y_i$が正解値$y_i$と同じ(=正しく予測できた)場合は$(y_i-\hat y_i)$が0となり、更新値も0になるため、更新は事実上行われない。<br>
予測値が正解値と違う(=正しく予測できなかった)場合は$(y_i-\hat y_i)$が非0となり、更新が行われる。<br>
学習率の注意点として、学習アルゴリズム内で更新されないパラメータ(**ハイパーパラメータ**)であるため、人間側で決める必要がある。<br>
学習率が小さすぎると重みの更新がなかなか進まず、大きすぎると重みが収束から遠ざかりつづけてしまうことがあるため、良い感じの値を探索して見つける必要がある。<br>
### プログラムに適用する
プログラムに適用しやすくするために、式の活用手順を明確にする。<br>
1. 初めに、重みベクトル$w$とバイアスユニット$b$を初期化する(0か小さい乱数値)
2. 入力データ$x_{i}$(1行)ごとに総入力$z$を計算して、閾値関数$σ$から出力値(予測値)$\hat y_i$を得る
3. 予測値に基づいて重みとバイアスを更新する
4. 2と3の演算を全ての行で行う


注意点1：出力は1行の入力ごとに得るため、場合によっては、重みとバイアスの更新も1行の出力が得られた時に1行分行う(今回は数式通り)。<br>
注意点2：バイアスは行列ではないため、行ごとのバイアスの更新値は全て一つのバイアス$b$に足される。<br>
注意点3：1つの重みを何度も更新するため、全ての行列の演算は繰り返し行われる(例えば、1000回とか)。

```python
%pip install numpy matplotlib scikit-learn

# -*- coding: utf-8 -*-
import numpy as np

class Perceptron:
    """
    # パラメータ
    **X**       :入力データ<br>
    **y**       :正解データ<br>
    **alpha**   :学習率<br>
    **n_iters** :行列の重み更新の繰り返し回数<br>
    **weights** :重みベクトル(必須ではない)<br>
    **bias**    :バイアス(必須ではない)
    """
    def __init__(self, X, y, alpha=0.01, n_iters=1000, weights=None, bias=None):
        self.X_         = X
        self.y_         = y
        self.alpha_     = alpha
        self.n_iters_   = n_iters
        self.weights_   = weights
        self.bias_      = bias
        
    def fit(self, get_accuracy=False):
        """
        パーセプトロンモデルを入力データに適合させる
        """
        row, col = self.X_.shape        # row:(データ数=サンプル数)、col:列数(特徴量の数=次元数)
        
        # パラメータ(重みとバイアス)の初期化
        if self.weights_ is None:
            self.weights_ = np.zeros(col)   # 注意：数式では重みベクトルは1行col列として考えているが、プログラム内部ではcol行1列となっている
        if self.bias_    is None:
            self.bias_    = 0
                    
        # 重みとバイアスを求める
        for _ in range(self.n_iters_):  # 行列全体の計算をn_iters回繰り返す
            
            y_pred = self.predict(self.X_)                          # 入力データのを用いて、入力行ごとのクラスラベル(列)を予測する
            error = self.y_ - y_pred                               # 予測値と正解値の差を求める(正しい予測の場合は0、誤った予測の場合は非0)
            
            delta_weights = self.alpha_ * np.dot(error.T, self.X_)   # 予測値と入力データの積の和を学習率で調整した値を、重みの更新量とする
            delta_bias    = self.alpha_ * sum(error)               # 予測値と正解値の差の和を学習率で調整した値を、バイアスの更新量とする
                        
            self.weights_ += delta_weights                          # 重みを更新する
            self.bias_    += delta_bias                             # バイアスを更新する
                        
        if get_accuracy:
            return self.accuracy(self.y_, self.predict(self.X_))
        
    def predict(self, X):
        """総入力zを基に、決定関数を用いてクラスラベル(0か1)を予測する
        """
            
        z = self.net_input(X)
        y_pred = self.threshold(z) 
        return y_pred
    
    def net_input(self, X):
        """総入力zを計算する
        """
        return np.dot(X, self.weights_) + self.bias_
    
    def threshold(self, z):
        """閾値関数<br>
        与えられた総入力zを基に、0か1を返す
        """
        #return 1 if z >= 0 else 0      # accuracy実行時に行列が全て代入されるため、その場合はzは1次元の配列になり、このコードでは対応できない.
        return np.where(z >= 0, 1, 0)   # 行列にも対応したコード
    
    def accuracy(self, y_true, y_pred):
        """正解率を計算する
        """
        return sum(y_true == y_pred) / len(y_true)
    
    def plot_dicision_boundary(self, resolution=0.01):
        """決定境界をプロットする<br>
        resolution:決定境界をプロットするためのグリッドの解像度(グリッドの点と点の距離)
        """
        import matplotlib.pyplot as plt
        
        # 決定境界をプロットするためのグリッドを作成
        x_min, x_max = self.X_[:, 0].min() - 1, self.X_[:, 0].max() + 1
        y_min, y_max = self.X_[:, 1].min() - 1, self.X_[:, 1].max() + 1
        
        # 表示範囲を設定
        plt.xlim(x_min, x_max)
        plt.ylim(y_min, y_max)
        
        # x_minとx_maxの地点で、予測値が変わるyの値を求める(2分探索)
        y_border_upper_x_min = y_border_upper_x_max = y_max
        y_border_lower_x_min = y_border_lower_x_max = y_min
        y_middle_x_min = y_middle_x_max = (y_max + y_min) / 2
        for _ in range(int(1/resolution)):
            y_pred_x_min = self.predict([x_min, y_middle_x_min])
            y_pred_x_max = self.predict([x_max, y_middle_x_max])
            
            if y_pred_x_min == 0:
                y_border_lower_x_min = y_middle_x_min
            else:
                y_border_upper_x_min = y_middle_x_min
                
            if y_pred_x_max == 0:
                y_border_lower_x_max = y_middle_x_max
            else:
                y_border_upper_x_max = y_middle_x_max
                
            y_middle_x_min = (y_border_upper_x_min + y_border_lower_x_min) / 2
            y_middle_x_max = (y_border_upper_x_max + y_border_lower_x_max) / 2
            
        # x_minとx_maxの地点で、予測値が変わるyの値を結ぶ直線を決定境界とする
        slope = (y_border_upper_x_max - y_border_upper_x_min) / (x_max - x_min)
        intercept = y_border_upper_x_min - slope * x_min
        x = np.array([x_min-1, x_max+1])
        y = slope * x + intercept
    
        # 決定境界近似曲線をプロット、境界で色分け
        plt.plot(x, y, color='black', linestyle='-', label=f'決定境界近似直線: y = {slope:.2f}x + {intercept:.2f}')
        plt.fill_between(x, y, y_min-1, color='blue', alpha=0.2, label='Class 0 領域')
        plt.fill_between(x, y, y_max+1, color='red' , alpha=0.2, label='Class 1 領域')
        
        # 正解値が0と1のデータ点を異なるマーカー、予測値が0と1のデータ点を異なる色でプロット
        cls0_collect = self.X_[np.logical_and(self.predict(self.X_) == 0, self.y_ == 0)]
        cls1_collect = self.X_[np.logical_and(self.predict(self.X_) == 1, self.y_ == 1)]
        cls1_error   = self.X_[np.logical_and(self.predict(self.X_) == 1, self.y_ == 0)]
        cls0_error   = self.X_[np.logical_and(self.predict(self.X_) == 0, self.y_ == 1)]
        plt.scatter(cls0_collect[:, 0], cls0_collect[:, 1], c='blue', edgecolors='k', marker='o', label='Class 0 (Predicted True)')
        plt.scatter(cls1_collect[:, 0], cls1_collect[:, 1], c='red' , edgecolors='k', marker='s', label='Class 1 (Predicted True)')
        plt.scatter(cls1_error[:, 0]  , cls1_error[:, 1]  , c='cyan' , edgecolors='k', marker='o', label='Class 0 (Predicted False)')
        plt.scatter(cls0_error[:, 0]  , cls0_error[:, 1]  , c='orange', edgecolors='k', marker='s', label='Class 1 (Predicted False)')
                
        # 軸ラベル、タイトル、凡例の設定
        plt.xlabel(xlabel='sepal length (cm)', fontdict={'family':'MS Gothic'})
        plt.ylabel(ylabel='petal length (cm)', fontdict={'family':'MS Gothic'})
        plt.title(label='パーセプトロンの決定領域と予測結果' , fontdict={'family':'MS Gothic'})
        plt.legend(prop={'family':'MS Gothic', 'size':7}, loc='lower right', )
        
        plt.show()
```

# ADALINE=ADAptive LInear NEuron
パーセプトロンアルゴリズムが発表された数年後に発表された。<br>
パーセプトロンと同じ点は、入力データから0か1の予測をする2値分類アルゴリズムであること。<br>
パーセプトロンと違う点は、**境界の決め方が違う**こと。<br>
発想としては、「**パーセプトロンは離散的な予測値で学習するから、逆に連続値を用いて学習するモデルを考案した**」と解釈できる。
### ADALINEの構造
入力を1つ与えると、出力が1つ得られる。<br>
出力の種類は、0か1の2種類あり、入力の信号(離連続値)がニューロンに与えられると、0か1(離散値)が出力される。<br>
ニューロンには閾値が存在し、入力値が閾値以上ならば出力は1、閾値未満なら出力は0となる。<br>
入力に対する出力に正解がある前提で、正解値と予測値の差を連続値の**誤差**と考え、誤差を用いた損失関数の値である**損失を最小化**することを目指す。
### ADALINEの原理解説
パーセプトロンの原理の解説と同様に、入力値ベクトルと正解値ベクトルが与えられるとする。
##### 活性化関数$σ$
パーセプトロンの閾値関数とADALINEの活性化関数は、どちらも**入力に基づいた出力**だが、大きな違いがある。<br>
パーセプトロンの閾値関数とは異なり、ADALINEの活性化関数は**微分によって勾配(=微分値)を求めることができる**。<br>
**なぜ微分できる必要があるのか**を考える。<br>
連続な曲線である損失関数の損失を限りなく最小に近付けるためには、**曲線のどの方向(左右)に進むと損失が最小に近づくかを常に算出できる必要がある**。<br>
これは、**損失関数の勾配(坂)を下る必要がある**ことを意味している。<br>
損失関数の勾配を求めるには、**その中で使われている式(活性化関数)もまた、微分可能でなくてはいけない**。<br>
パーセプトロンの閾値関数は連続ではないため、微分が出来ない。<br>
活性化関数は、微分が出来ない閾値関数の代わりとして、損失関数への入力を求める役割があると捉えられる。<br>
また、閾値関数は予測値を求めるときのみ用いられることから、**閾値関数の役割は予測値に変換することに限定される**と考えられる(パラメータの更新に用いるかは別問題ということ)。<br>
ちなみに、現在では活性化関数は複数存在するが、全てADALINEが基礎となっていると見なすことができる。
$$
\begin{gather}
活性化関数(最もシンプル)\\
\sigma(z_m)=z_m=w\cdot x_m+b
\end{gather}
$$
##### 損失関数$L$
損失関数には色んな種類があるが、今回は**平均二乗誤差=MSE**を用いる場合を考える。<br>
平均二乗誤差は、正解値$y$と活性化関数からの出力$σ(z)$の2乗を入力数で割った値で、2乗の部分を平方根にすると式全体が距離になる。<br>
平方根がない形が使われているのは、こっちの方が計算が楽だからと聞いたことがある。<br>
$$
\begin{gather}
損失関数(平均二乗誤差, MSE)\\
L(w,b) = \frac{1}{n}\sum_m(y_m-\sigma (z_m))^2
\end{gather}
$$
パラメータの更新には、損失関数の**勾配**(=微分値)「$\nabla_wL(w,b)$」と「$\nabla_bL(w,b)$」と学習率$\alpha$を用いる。<br>
**勾配が大きくなるほど更新値は大きくなる**ということは、誤差が大きいほどパラメータの更新が大きくなり、誤差が小さいほどパラメータの更新は小さくなることを意味する。<br>
更新式にはマイナスがついているため、勾配を下る方向、つまり、**損失関数の値を最小値に近付ける方向にパラメータが更新される**。<br>
このように、関数の最小値や最大値を勾配を用いて見つける手法を**勾配降下法**と呼び、今回使うのはそのうちの一つということになる。
$$
\begin{gather}
重みベクトルとバイアスの更新式\\
w\leftarrow w+\Delta w\\
b\leftarrow b+\Delta b\ \\
\\
更新値の計算式(勾配降下法を用いる)\\
\Delta w=-\alpha \nabla_w L(w,b), \ \ \Delta b=-\alpha \nabla_b L(w,b)\\
\end{gather}
$$
損失関数の偏微分を示す。
$\nabla_wL(w,b)$で重みベクトルの勾配、$\nabla_bL(w,b)$でバイアスの勾配が求まる(途中式はめんどいから省略)。
$$
\begin{gather}
偏導関数(\nabla)の式\\
\nabla_w L(w,b) = -\frac{2}{n} \sum_m(y_m-\sigma(z_m))x_{mn}\\
\nabla_b L(w,b) = -\frac{2}{n} \sum_m(y_m-\sigma(z_m))
\end{gather}
$$
$w$についての偏導関数の式がちょっと曖昧で正確かどうか怪しいが、正解値と活性化関数からの誤差のベクトルと入力の各列の内積から、ベクトルが算出される。
### ADALINEの実装
##### データ構造
$i$行$j$列の入力データと、行ごとの正解データが与えられるとする。<br>
重みベクトル$w$とバイアス$b$を学習で更新したい。
$$

\begin{gather}
事前に与えられるデータ \\
入力データ\ 
x= 
\left[
\begin{array}{}
  x_{11}&x_{12}&\cdots&x_{1j}\\
  x_{21}&x_{22}&\cdots&x_{2j}\\
  \vdots&\vdots&\ddots&\vdots\\
  x_{i1}&x_{i2}&\cdots&x_{ij}
\end{array}
\right],\ \
正解データ\ y=\left[
\begin{array}{}
y_1\\
y_2\\
\vdots\\
y_i
\end{array}
\right]\\\\
学習で求めたい値\\
重みベクトル\ \ w=\left[
\begin{array}{}
w_1 & w_2 & \cdots & w_j
\end{array}
\right],\ \ 
バイアス\ b\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \
\end{gather}
$$
##### 学習則
総入力ベクトル$z$は、正解データ$y$と同じ形状になり、活性化関数に通したベクトルも同じ形状になる。<br>
$error$を正解データ$y$と活性化関数の出力ベクトル$σ(z)$のから得られた**損失ベクトル**とすると、正解データ$y$と同じ形状になる。<br>
$\nabla_wL$について、$1$行$i$列である$error^T$と$i$行$j$列である入力ベクトル$x$のドット積は$1$行$j$列となり、これが重みベクトル$w$と同じ形状になる。
$$
\begin{gather}
総入力ベクトル\\
z=
\left[
\begin{array}{}
w \cdot x_1 + b\\
w \cdot x_2 + b\\
\vdots\\
w \cdot x_i + b
\end{array}
\right]
=
\left[
\begin{array}{}
z_1\\
z_2\\
\vdots\\
z_i
\end{array}
\right]\\ \\
活性化関数ベクトル\\
\sigma(z)=z\\ \\
重みベクトルとバイアスの更新式\\
w\leftarrow w+\Delta w\\
b\leftarrow b+\Delta b\ \\\\
更新値の計算式(\alpha は任意の値)\\
\Delta w= - \alpha \nabla_wL(z)\\
\Delta b= - \alpha \nabla_bL(z) \\ \\
偏微分(nはサンプル数)\\
error=(y-\sigma(z))
=
\left[
\begin{array}{}
error_1\\
error_2\\
\vdots\\
error_i
\end{array}
\right]\\
\nabla_wL(z)=-\frac{2}{n}(error^T \cdot x)\\
\nabla_bL(z)=-\frac{2}{n}\sum_i error_i
\end{gather}
$$
予測値を得る際は、活性化関数を計算した後、閾値関数に通す。
### プログラムに適用する
パーセプトロンとの違いは、更新値の計算式が損失関数を用いた勾配になっていることと、誤差の更新に閾値関数ではなく活性化関数が用いられていること。<br>
予測値を得る時は、パーセプトロンと同様に閾値関数を用いる。

```python
# -*- coding: utf-8 -*-
import numpy as np

class ADALINE:
    """
    # パラメータ
    **X**       :入力データ<br>
    **y**       :正解データ<br>
    **alpha**   :学習率<br>
    **n_iters** :行列の重み更新の繰り返し回数<br>
    **weights** :重みベクトル(必須ではない)<br>
    **bias**    :バイアス(必須ではない)
    """
    def __init__(self, X, y, alpha=0.01, n_iters=1000, weights=None, bias=None):
        self.X_         = X
        self.y_         = y
        self.alpha_     = alpha
        self.n_iters_   = n_iters
        self.weights_   = weights
        self.bias_      = bias
        
    def fit(self, get_accuracy=False):
        """
        ADALINEモデルを入力データに適合させる
        """
        row, col = self.X_.shape        # row:(データ数=サンプル数)、col:列数(特徴量の数=次元数)
        
        # パラメータ(重みとバイアス)の初期化
        if self.weights_ is None:
            self.weights_ = np.zeros(col)   # 注意：数式では重みベクトルは1行col列として考えているが、プログラム内部ではcol行1列となっている
        if self.bias_    is None:
            self.bias_    = 0
                    
        # 重みとバイアスを求める
        for _ in range(self.n_iters_):  # 行列全体の計算をn_iters回繰り返す
            
            z = self.net_input(self.X_)                                     # 総入力を計算しておく
            nabla_wL, nabla_bL = self.nabla_objective(z)  # 目的関数(損失関数)の偏微分値を計算
            
            delta_weights = - self.alpha_ * nabla_wL                        # 重みの更新量
            delta_bias    = - self.alpha_ * nabla_bL                        # バイアスの更新量
            self.weights_ += delta_weights                                  # 重みを更新する
            self.bias_    += delta_bias                                     # バイアスを更新する
                        
        if get_accuracy:
            return self.accuracy(self.y_, self.predict(self.X_))
        
    def predict(self, X):
        """総入力zを活性化関数に通して、決定関数を用いてクラスラベル(0か1)を予測する
        """
            
        z = self.net_input(X)
        activated_z = self.activation(z)
        y_pred = self.threshold(activated_z)
        
        return y_pred
    
    def net_input(self, X):
        """総入力zを計算する
        """
        return np.dot(X, self.weights_) + self.bias_
    
    def threshold(self, z):
        """決定関数<br>
        与えられた総入力zを基に、0か1を返す
        """
        #return 1 if z >= 0 else 0      # accuracy実行時に行列が全て代入されるため、その場合はzは1次元の配列になり、このコードでは対応できない.
        return np.where(z >= 0.5, 1, 0)   # 行列にも対応したコード
    
    def activation(self, z):
        """活性化関数<br>
        与えられた総入力zを基に、値を返す
        """
        return z
    
    def nabla_objective(self, z):
        """目的関数の偏微分<br>
        ADALINEの目的関数は損失関数<br>
        損失関数の偏微分値を返す
        数式上は重みベクトルとバイアスを代入するが、このプログラムでは総入力を入れる
        """
        error = self.y_ - self.activation(z)                    # i行1列
        nabla_wL = -(2 / len(self.y_)) * np.dot(error.T, self.X_) # 1行i列 × i行j列 = 1行j列
        nabla_bL = -(2 / len(self.y_)) * sum(error)
        
        return nabla_wL, nabla_bL
        
    
    def accuracy(self, y_true, y_pred):
        """正解率を計算する
        """
        return sum(y_true == y_pred) / len(y_true)
    
    def plot_dicision_boundary(self, resolution=0.01):
        """決定境界をプロットする<br>
        resolution:決定境界をプロットするためのグリッドの解像度(グリッドの点と点の距離)
        """
        import matplotlib.pyplot as plt
        
        # 決定境界をプロットするためのグリッドを作成
        x_min, x_max = self.X_[:, 0].min() - 1, self.X_[:, 0].max() + 1
        y_min, y_max = self.X_[:, 1].min() - 1, self.X_[:, 1].max() + 1
        
        # 表示範囲を設定
        plt.xlim(x_min, x_max)
        plt.ylim(y_min, y_max)
        
        # x_minとx_maxの地点で、予測値が変わるyの値を求める(2分探索)
        y_border_upper_x_min = y_border_upper_x_max = y_max
        y_border_lower_x_min = y_border_lower_x_max = y_min
        y_middle_x_min = y_middle_x_max = (y_max + y_min) / 2
        for _ in range(int(1/resolution)):
            y_pred_x_min = self.predict([x_min, y_middle_x_min])
            y_pred_x_max = self.predict([x_max, y_middle_x_max])
            
            if y_pred_x_min == 0:
                y_border_lower_x_min = y_middle_x_min
            else:
                y_border_upper_x_min = y_middle_x_min
                
            if y_pred_x_max == 0:
                y_border_lower_x_max = y_middle_x_max
            else:
                y_border_upper_x_max = y_middle_x_max
                
            y_middle_x_min = (y_border_upper_x_min + y_border_lower_x_min) / 2
            y_middle_x_max = (y_border_upper_x_max + y_border_lower_x_max) / 2
            
        # x_minとx_maxの地点で、予測値が変わるyの値を結ぶ直線を決定境界とする
        slope = (y_border_upper_x_max - y_border_upper_x_min) / (x_max - x_min)
        intercept = y_border_upper_x_min - slope * x_min
        x = np.array([x_min-1, x_max+1])
        y = slope * x + intercept
    
        # 決定境界近似曲線をプロット、境界で色分け
        plt.plot(x, y, color='black', linestyle='-', label=f'決定境界近似')
        plt.fill_between(x, y, y_min-1, color='blue', alpha=0.2, label='Class 0 領域')
        plt.fill_between(x, y, y_max+1, color='red' , alpha=0.2, label='Class 1 領域')
        
        # 正解値が0と1のデータ点を異なるマーカー、予測値が0と1のデータ点を異なる色でプロット
        cls0_collect = self.X_[np.logical_and(self.predict(self.X_) == 0, self.y_ == 0)]
        cls1_collect = self.X_[np.logical_and(self.predict(self.X_) == 1, self.y_ == 1)]
        cls1_error   = self.X_[np.logical_and(self.predict(self.X_) == 1, self.y_ == 0)]
        cls0_error   = self.X_[np.logical_and(self.predict(self.X_) == 0, self.y_ == 1)]
        plt.scatter(cls0_collect[:, 0], cls0_collect[:, 1], c='blue', edgecolors='k', marker='o', label='Class 0 (予測成功)')
        plt.scatter(cls1_collect[:, 0], cls1_collect[:, 1], c='red' , edgecolors='k', marker='s', label='Class 1 (予測成功)')
        plt.scatter(cls1_error[:, 0]  , cls1_error[:, 1]  , c='cyan' , edgecolors='k', marker='o', label='Class 0 (予測失敗)')
        plt.scatter(cls0_error[:, 0]  , cls0_error[:, 1]  , c='orange', edgecolors='k', marker='s', label='Class 1 (予測失敗)')
                
        # 軸ラベル、タイトル、凡例の設定
        plt.xlabel(xlabel='sepal length (cm)', fontdict={'family':'MS Gothic'})
        plt.ylabel(ylabel='petal length (cm)', fontdict={'family':'MS Gothic'})
        plt.title(label='ADALINEの決定領域と予測結果' , fontdict={'family':'MS Gothic'})
        plt.legend(prop={'family':'MS Gothic', 'size':7}, loc='lower right', )
        
        plt.show()
```

# パーセプトロンとADALINEの比較
パーセプトロンもADALINEも、重みは特徴量に対する注目の度合いを表していて、**重みの絶対値が大きいほど注目すべき特徴**であることを意味すると考えられる。<br>
パーセプトロンは、誤分類の有無という離散値のみを用いて学習するため、「間違えた」という事実に基づいてパラメータを更新すると捉えられる。<br>
ADALINEは、予測値と正解値の誤差(連続値)を用いて学習するため、「どれくらい間違えたか」というスコアに基づいて学習が進むと捉えられる。<br>
ここで、1つの直線で線引きできない問題を考える。<br>
パーセプトロンは、線形分類問題では、必ず一定の線引きができるが、非線形分類問題では、「間違えた」という結果に影響されつづけることが原因で、一定の判断基準を求められない。<br>
一方ADALINEは、「誤差が最も少ない線引き」を目指すため、極端な外れ値に引っ張られてしまうことがある反面、どんなデータでも一定の判断基準を求められる。<br>
現実の問題は非線形分類問題の方が多く、パーセプトロンよりはADALINEの方が使い道が多いため、ADALINEはパーセプトロンの改良版と見なすことができる。
