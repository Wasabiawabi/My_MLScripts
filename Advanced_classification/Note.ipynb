# データの前処理とパラメータの調整
### 勾配降下法の効率化-標準化
スケーリングの一種で、**データの平均値を0、標準偏差を1にする**計算です。<br>
標準偏差が1なので、2乗した分散も1になります。
$$
\begin{gather}
入力データ\

x=

\left[

\begin{array}{}

  x_{11}&x_{12}&\cdots&x_{1j}\\

  x_{21}&x_{22}&\cdots&x_{2j}\\

  \vdots&\vdots&\ddots&\vdots\\

  x_{i1}&x_{i2}&\cdots&x_{ij}

\end{array}

\right], \ 
x_j = \left[

\begin{array}{}

x_{1j}\\

x_{2j}\\

\vdots\\

x_{ij}

\end{array}

\right] \ \ \\ \\
標準化 \
x_j'=\frac{x_j-\mu_j}{\sigma_j}\ \ \ (\mu_jはj列目の平均値,\ \sigma_jはj列目の標準偏差)
\end{gather}
$$
計算内容は$x_j- \mu_j$の平行移動と$/\sigma_j$のスケーリングのみなので、**データの相対的な構造は変わらない**という特徴があります。<br>
最小値と最大値に指定がない場合や、外れ値が存在する場合に有効です。
##### もう一つの手法
もう一つのスケーリングの手法に、**正規化**があり、データを0~1の間に収めます。<br>
機械学習でデータの前処理を行う場合は、一般的に標準化の方がよく使われます。
### 過学習の回避-L2正則化
モデルを学習する時、パラメータが訓練データに過剰適合する**過学習**に陥り、**訓練データの予測性能に比べて未知データの予測性能がかなり低くなる**ことがあります。<br>
過学習を緩和するには、**モデルの出力がパラメータに依存しすぎないようにする必要**があります。<br>
損失関数に**正則化項=ペナルティ項**を加え、ペナルティ項まで含めた値を損失と考えるようにします。<br>
損失を最小化するという考え方の元、**損失関数に加え、ペナルティ項の最小化を目指すようにモデルが学習する**ようになります。
$$
\begin{gather}
重みベクトル\\
w=\left[
\begin{array}{}
w_1 & w_2 & \cdots & w_j
\end{array}
\right]\\\\
L2正則化項(\lambda は任意の値)\\
\frac{\lambda}{n}||w||^2_2=\frac{\lambda}{n}\sum_jw_j^2\\ \\
損失関数にL2正則化を追加\\
L(w,b)\leftarrow L(w,b)+\frac{\lambda}{n}||w||^2_2\\\\
偏微分した場合\\
\nabla_wL(w,b)\leftarrow\nabla_wL(w,b)+\frac{2\lambda}{n}w\\
\nabla_bL(w,b)\leftarrow\nabla_bL(w,b)+0
\end{gather}
$$
L2正則化を用いて学習した場合、**パラメータは極端に0に近づきます**。
$\lambda$を大きくするほど正則化項の影響が大きくなり、パラメータは0に近い値になりやすくなります。
正則化は入力データに基づいて学習される重みを用いる関係上、**特徴量のスケールが違うと計算が上手く行きません。**<br>
正則化をする場合は、特徴量のスケールを合わせられる**標準化をしておくと安心**です。
##### 他の正則化
L2正則化と同じくらい有名な手法として、**L1正則化**があります。<br>
L2正則化を用いて算出された重みは0に極端に近いことが多いですが、L1正則化の場合は**一部の重みが0になります**。<br>
これには、**スパース化=特徴量選択**という効果があり、出力値の予測対する影響が少ない特徴量をなくすことができます。<br>
(数学的な証明もあるようですが、ここでは飛ばします。)
$$
\begin{gather}
L1正則化\\
||w||_1=\sum_j|w_j|
\end{gather}
$$
また、0でない重みの数を用いる**L0正則化**もあります。<br>
計算量がとても多くなりやすいためあまり使われることはなく、**微分が不可能**なため、機械学習ではほとんど使うことが出来ません。<br>
ディープラーニングでは、正則化の近似として間接的に使われることが多いようです。<br>

```python
# -*- coding : utf-8 -*-
import numpy as np

def Standarize(X):
    """標準化

    Args:
        X (ベクトル): 入力データ
        
    Returns:
        X (ベクトル): 標準化後の入力データ
    """
    return (X - np.mean(X)) / np.std(X)

def L2Regularization(w, a):
    """L2正則化<br>
    実用を考えるならば、モデルの中に組み込む必要がある.<br>
    今回は、関数として書いてみただけ.
    
    Args:
        X (ベクトル): 重みベクトル
        a (float): 正則化係数
        
    Returns:
        w (ベクトル): 正規化後の重みベクトル
    """
    return (a / len(w)) * sum(w ** 2)
```

# アルゴリズムの改良
### 活性化関数の改良(ロジスティック回帰)
##### 出力確率も知りたい
二値分類では出力が離散値(0か1)で得られますが、**どれくらいの確率でその出力になるのか**を求めたいことが往々にしてあります。<br>
患者が特定の病気に感染する確率や、顧客が特定の商品を購入する確率など、様々な例が挙げられます。<br>
**活性化関数を確率の表現ができる関数に変える**ことが出来れば、出力の確率を確かめることが出来るようになります。<br>
また、出力の確率をそのまま損失関数に用いる損失とすれば、**モデルの学習も問題なく行えます**。<br>
確率の表現ができる関数は**シグモイド関数**と呼ばれ、シグモイド関数を用いた分類アルゴリズムは**ロジスティック回帰**と呼ばれます。
##### シグモイド関数
ここで、関数によって得たい確率を$p$と置きます。<br>
モデルのパラメータを用いた式から、**正事象**の確率$p$を得たいです。<br>
モデルで用いられる全てのパラメータを用いている総入力$z$を用いることを考えます。
$$
\begin{gather}
式のイメージ\\
p=(総入力zを用いた式)
\end{gather}
$$
総入力$z$から**正事象**の起こりやすさを求めることを考えます。<br>
正事象の起こりやすさは**オッズ比**と呼ばれます。<br>
計算の都合上、オッズ比の対数をとった**対数オッズ比**を用います。
$$
\begin{gather}
オッズ比OR\\
OR=\frac{p}{1-p}\\\\
対数オッズ比\\
log(OR)=\log_e\frac{p}{1-p}
\end{gather}
$$
総入力$z$が対数オッズ比を直接表していると仮定して、正事象の確率$p$の計算式を求めます。
$$
\begin{gather}
log_e\frac{p}{1-p}=z\\\\
対数の変換(\log_ac=b \iff a^b=c)\\
\frac{p}{1-p}=e^z\\\\
\frac{1-p}{p}=\frac{1}{e^z}\\\\
\frac{1}{p}-1=e^{-z}\\\\
\frac{1}{p}=1+e^{-z}\\\\
p=\frac{1}{1+e^{-z}}
\end{gather}
$$
総入力$z$を任意の実数に対応させれば、シグモイド関数の出来上がりです。<br>
シグモイド関数は正事象の確率$p$を表すことができています。
$$
\begin{gather}
シグモイド関数(-\infty \le z \le \infty)\\
\sigma(z)=\frac{1}{1+e^{-z}}
\end{gather}
$$
##### シグモイドじゃない関数を考えてみる
ちなみに、便宜上付け加えた対数がないと、総入力$z$の対応範囲は$0\le z$となってしまいます。
$$
\begin{gather}
対数オッズ比を用いない場合\\
\frac{p}{1-p}=z\\\\
\frac{1-p}{p}=\frac{1}{p}-1=z\\\\
\frac{1}{p}=1+z\\\\
p=\frac{1}{1+z}\\\\
シグモイドじゃない関数(0 \le z)\\
\sigma(z)=\frac{1}{1+z}
\end{gather}
$$
##### シグモイド関数を微分する
$$
\begin{gather}
シグモイド関数を微分\\
\sigma'(z)=\frac{e^{-z}}{(1+e^{-z})^2}\\\\
\frac{e^{-z}}{(1+e^{-z})^2}=\frac{e^{-z}}{1+e^{-z}}\cdot\frac{1}{1+e^{-z}}=
\frac{1}{1+e^{-z}}\cdot\Big(\frac{1+e^{-z}}{1+e^{-z}}-\frac{1}{1+e^{-z}}\Big)\\\\
=\sigma(z)\cdot\big(1-\sigma(z)\big)
\end{gather}
$$
シグモイド関数$\sigma(z)$は正事象の確率$p$を表しているため、$\sigma(z)$は$p$に置き換えることができ、これは**ベルヌーイ分布の分散**と同じ形になります。
$$
\sigma(z)\cdot\big(1-\sigma(z)\big)=p\cdot(1-p)
$$
##### 補足説明
ロジスティック回帰は、「回帰」という言葉がありますが、分類問題に対応したアルゴリズムになります。<br>
回帰アルゴリズムと分類アルゴリズムの違いは出力の違いにあり、**回帰アルゴリズムは連続値を直接出力とし、分類アルゴリズムは出力を離散値に変換**します。<br>
ロジスティック回帰は、シグモイド関数から来る連続値を活用しているため、**回帰アルゴリズムと間違えられやすい**です。<br>
しかし、**最終的に予測したいのは**、例えば、患者が病気に感染しているか(1)いないか(0)といった**離散値**です。<br>
シグモイド関数を用いるのは、最終的な離散値の出力を得る前に、その**離散値の出力になる確率を事前に確認したい**ためです。<br>
一方の**回帰問題は**、住宅の価格予測や次の日の気温予測など、**出力そのものが連続値**になります。<br>

まとめると、「**回帰アルゴリズムは連続値を予測するアルゴリズム、分類アルゴリズムは離散値を予測するアルゴリズム。<br>
ロジスティック回帰は離散値の予測確率を確認するために、活性化関数としてシグモイド関数を使っている**」となります。

### 確率に対応した損失関数(交差エントロピー)
確率を出力するシグモイド関数を用いてパラメータを更新したいため、確率を用いた損失関数を定義することを考えます。<br>
正解値$y$と予測値$\sigma(z)$が近くなるほど出力値が0に近づくような関数であれば、損失関数として使うことができます。<br>
**交差エントロピー**は、これらの要求を満たすことが出来ます。
$$
\begin{gather}
交差エントロピー\\
H(p,q)=-\sum_xp(x)\log \big(q(x)\big)
\end{gather}
$$
$p(x)$が1で正解、0で不正解と考えると、予測値$q(x)$が正解に近づくほど$H(p,q)$は0に近づくため、今回の損失関数に求められる性質を持っていることが確認できます。
ただし、これは確率を関数で表した一般分布で用いられる式です。<br>
機械学習の2値分類で使いたい場合は、$p(x)\in\{0,1\}$という条件に合わせて展開する必要があります。<br>
展開した交差エントロピーは、**2値交差エントロピーとなります**。
$$
\begin{gather}
2値交差エントロピー\\
L(p,q)=-\sum_x\Big(p(x)\log\big(q(x)\big)+\big(1-p(x)\big)\log\big(1-q(x)\big)\Big)\ \ \ \ \ \big(p(x)\in\{0,1\},\ \ 0<q(x)<1\big)
\end{gather}
$$
正解値が$p(x)=1$の場合は$p(x)$がある項のみが働き、$1-p(x)$がある項は0になります。<br>
正解値が$p(x)=0$の場合は$p(x)$がある項は0になり、$1-p(x)$がある項が働きます。<br>
これならば、正解値が2値の場合に適用できます。
##### モデル学習用の定義
$p(x)$は1入力の正解値、$q(x)$は1入力の予測値に当たるので、$p(x)=y_i$、$q(x)=\sigma(z_i)$と置き換えることができます。
$$
L(y,\sigma(z);w,b)=-\sum_i\Big(y_i\log\big(\sigma(z_i)\big)+\big(1-y_i)\log\big(1-\sigma(z_i)\big)\Big)\ \ \ \ \ \big(y_i\in\{0,1\},\ \ 0<\sigma(z_i)<1\big)
$$
また、モデルのパラメータの更新式に当てはめるために、偏微分を行います。
$$
\begin{gather}
&パラメータ更新式\\
&w\leftarrow w+\Delta w\\
&b\leftarrow b+\Delta b\\
&\Delta w=-\alpha \nabla_wL(y,\sigma(z);w,b)\\
&\Delta b=-\alpha \nabla_bL(y,\sigma(z);w,b)\\\\
&偏微分する\\
&\nabla_wL(y,\sigma(z);w,b)\\
&=\frac{\partial z}{\partial w}\cdot\frac{\partial \sigma(z)}{\partial z}\cdot\frac{\partial L}{\partial \sigma(z)}\\
&=\sum_i\bigg[\sum_jx_j\cdot \Big(\sigma(z)\cdot \big(1-\sigma(z) \big) \Big)\cdot -\Big(\frac{y}{\sigma(z)}-\frac{1-y}{1-\sigma(z)} \Big)\bigg]_i\\
&=\sum_i\bigg[-\sum_jx_j\Big\{y\big(1-\sigma(z)\big)-(1-y)\sigma(z)\Big\}\bigg]_i\\
&=\sum_i\bigg[\sum_jx_j\big(\sigma(z)-y)\bigg]_i\\\\
&バイアスはx_jがなくなった版\\
&\nabla_bL(y,\sigma(z);w,b)\\
&=\sum_i\bigg[-\Big(y\big(1-\sigma(z)\big)-(1-y)\sigma(z) \Big)\bigg]_i\\
&=\sum_i\Big[\sigma(z_i)-y_i\Big]_i
\end{gather}
$$
偏微分の式は、ニューラルネットワークの誤差逆伝播法に似ています。<br>
ニューラルネットワークの基本形は、ロジスティック回帰における、総入力$z$への線形変換と活性化関数$\sigma(z)$を何重にも連結したものです。<br>
ロジスティック回帰は、ニューラルネットワークの1層版と見なすことができます。
##### 実は交差エントロピーじゃなくて良い
微分可能かつ予測値が正解値に近づくほど更新する値が小さくなれば良いため、交差エントロピーではなく、平均二乗誤差を用いても問題なく計算できます。<br>
ただし、原理に沿った関数を使った方が、安定的な学習を担保しやすいため、交差エントロピーがよく使われます。

```python
# -*- coding: utf-8 -*-
import numpy as np

class LogisticRegression:
    """
    ADALINEをコピペして活性化関数と目的関数を変更
    # パラメータ
    **X**       :入力データ<br>
    **y**       :正解データ<br>
    **alpha**   :学習率<br>
    **n_iters** :行列の重み更新の繰り返し回数<br>
    **weights** :重みベクトル(必須ではない)<br>
    **bias**    :バイアス(必須ではない)
    """
    def __init__(self, X, y, alpha=0.01, n_iters=1000, weights=None, bias=None):
        self.X_         = X
        self.y_         = y
        self.alpha_     = alpha
        self.n_iters_   = n_iters
        self.weights_   = weights
        self.bias_      = bias
        
    def fit(self, get_accuracy=False):
        """
        ロジスティック回帰モデルを入力データに適合させる
        """
        row, col = self.X_.shape        # row:(データ数=サンプル数)、col:列数(特徴量の数=次元数)
        
        # パラメータ(重みとバイアス)の初期化
        if self.weights_ is None:
            self.weights_ = np.zeros(col)   # 注意：数式では重みベクトルは1行col列として考えているが、プログラム内部ではcol行1列となっている
        if self.bias_    is None:
            self.bias_    = 0
                    
        # 重みとバイアスを求める
        for _ in range(self.n_iters_):  # 行列全体の計算をn_iters回繰り返す
            
            z = self.net_input(self.X_)                                     # 総入力を計算しておく
            nabla_wL, nabla_bL = self.nabla_objective(z)    # 目的関数(損失関数)の偏微分値を計算
            
            delta_weights = - self.alpha_ * nabla_wL                        # 重みの更新量
            delta_bias    = - self.alpha_ * nabla_bL                        # バイアスの更新量
            self.weights_ += delta_weights                                  # 重みを更新する
            self.bias_    += delta_bias                                     # バイアスを更新する
                        
        if get_accuracy:
            return self.accuracy(self.y_, self.predict(self.X_))
        
    def predict(self, X):
        """総入力zを活性化関数に通して、決定関数を用いてクラスラベル(0か1)を予測する
        """
            
        z = self.net_input(X)
        activated_z = self.activation(z)
        y_pred = self.threshold(activated_z)
        
        return y_pred
    
    def net_input(self, X):
        """総入力zを計算する
        """
        return np.dot(X, self.weights_) + self.bias_
    
    def threshold(self, z):
        """決定関数<br>
        与えられた総入力zを基に、0か1を返す
        """
        #return 1 if z >= 0 else 0      # accuracy実行時に行列が全て代入されるため、その場合はzは1次元の配列になり、このコードでは対応できない.
        return np.where(z >= 0.5, 1, 0)   # 行列にも対応したコード
    
    def activation(self, z):
        """活性化関数<br>
        与えられた総入力zを基に、値を返す
        今回はシグモイド関数
        """
        return 1 / (1 + np.exp(-z))
    
    def nabla_objective(self, z):
        """目的関数の偏微分<br>
        ロジスティック回帰の目的関数は損失関数(2値交差エントロピー)<br>
        損失関数の偏微分値を返す
        数式上は活性化済みの総入力を代入するが、このプログラムでは総入力を入れる
        """
        error = self.activation(z) - self.y_
        nabla_wL = np.dot(error, self.X_)
        nabla_bL = np.sum(error)
        
        return nabla_wL, nabla_bL
        
    
    def accuracy(self, y_true, y_pred):
        """正解率を計算する
        """
        return sum(y_true == y_pred) / len(y_true)
    
    def plot_decision_boundary(model, X, y):# バグがあったため、変更
        import matplotlib.pyplot as plt
        # グリッド作成
        x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
        y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
        xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 200),
        np.linspace(y_min, y_max, 200)
        )
        # 予測
        Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
        Z = Z.reshape(xx.shape)

        # 図示
        plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
        plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm)
        plt.xlabel("sepal length")
        plt.ylabel("petal length")
        plt.title("自作ロジスティック回帰の決定境界", fontdict={'family':'MS Gothic'})
        plt.show()
        
        
if __name__ == "__main__":
    from sklearn.datasets import load_iris
    from pandas import DataFrame
    import Data_Adjustment
    
    # Irisデータセットの読み込み
    X, y = load_iris(return_X_y=True)
    
    # データの確認
    df = DataFrame(X, columns=load_iris().feature_names)
    df["target"] = y
    print(df.head())
    print(df.tail())
    
    # ロジスティック回帰モデルを適用できるようにするために、データの構造を変更する
    X = X[:-50, [0, 2]]                     # 入力データXを、sepalとpetalの長さとする(targetは3つあり、1つあたり50行ずつあるため、150行のうち前100行のみを選択)
    X = Data_Adjustment.Standardization(X)  # 標準化

    y = y[:-50]                             # 正解データyを、setosaとversicolorの2クラスにする(0と1の2値分類)
    
    # データ変更後のデータ確認
    df = DataFrame(X)
    df["target"] = y
    print(df.head())
    print(df.tail())
    
    # ロジスティック回帰モデルを学習
    model = LogisticRegression(X, y, alpha=0.01, n_iters=1000)
    acc = model.fit(get_accuracy=True)
    print(f"Accuracy: {acc}")
    
    # 決定境界のプロット
    model.plot_decision_boundary(X, y)
    
    print(model.weights_)
```

# 発展的なアルゴリズム(サポートベクターマシン=SVM)
SVMは、2クラス間の最も中間に当たる部分を決定領域にするという**マージン最大化問題**から着想を得ています。<br>
マージンの最大化を目指す理由は、全体の予測誤差が小さくなる傾向にあるためです。<br>
高度な数学の理論が使われていて、分類アルゴリズムの中で特に複雑ですが、その代わりに機械学習の分類アルゴリズムの中では最も重宝されています。<br>
**マージン**とは、決定境界から最も近いデータ点までの距離を指します。<br>
この最も近いデータ点を**サポートベクトル**と呼び、SVMでは、この距離（マージン）を最大化する決定境界を求めることになります。
##### スラック変数で非線形分類(ソフトマージン分類)に対応
SVMはパーセプトロンに似ています。<br>
パーセプトロンが非線形分類問題で決定境界を収束させられないのと同様に、初期のSVMも非線形分類問題が苦手です。<br>
スラック変数がSVMに導入され、スラック変数を制御するために**C変数**というものが導入されました。<br>
C変数は誤分類時のペナルティの大きさを決めることが出来ます。<br>
C変数が大きいほど誤分類に対するペナルティが大きくなりますが、マージンが小さくなる傾向にあります。<br>
C変数が小さいほど誤分類を許容する代わりに、マージンを大きくとるように学習するようになります。
##### 線形分離不可能な問題への対処
**カーネル法**を用いて、特徴量を高次元に写像することで高次元で分離可能にする手法です。<br>
簡単に言えば、**既にある特徴量から新しい特徴量を作り**、1つ上の次元でデータの決定領域を計算します。<br>
例えば、2次元で線引きできなくても、ガウスカーネルで2つの特徴量から3つめの特徴量を作れば、3次元で分離できる面が求められます。<br>
しかし、新しい特徴量を算出するのには高い計算コストがかかるため、**カーネルトリック**と呼ばれる手法を用いて内積のみを求めることで、計算を軽くします。<br>
**RBFカーネル=動径基底関数カーネル=ガウスカーネル**は次元数は無限に増やすことができ、これが最も広く使われているようです。
##### SVMの使用上の注意
データの前処理として、必ず標準化をする必要があります。<br>
これは、特徴量のスケールが違うとデータの距離の計算が狂うためです。<br>

```python
# -*- coding : utf-8 -*-
import numpy as np
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

# Irisデータセットの読み込み
X, y = load_iris(return_X_y=True)

# データの確認
df = pd.DataFrame(X, columns=load_iris().feature_names)
df["target"] = y
print(df.head())
print(df.tail())

# データの構造を変更する
X = X[:100, [0, 2]]    # 入力データXを、sepalとpetalの長さとする(targetは3つあり、1つあたり50行ずつあるため、150行のうち前100行のみを選択)
y = y[:100]            # 正解データyを、setosaとversicolorの2クラスにする(0と1の2値分類)

# データ変更後のデータ確認
df = pd.DataFrame(X, columns=["sepal length", "petal length"])
df["target"] = y
print(df.head())
print(df.tail())

# SVMモデルを学習
model = SVC(kernel="linear", C=1.0)
model.fit(X, y)

# 決定境界のプロット
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm)
plt.xlabel("sepal length")
plt.ylabel("petal length")
plt.title("SVMの決定境界", fontdict={'family':'MS Gothic'})
plt.show()
```
