# データの前処理とパラメータの調整
### 勾配降下法の効率化-標準化
スケーリングの一種で、**データの平均値を0、標準偏差を1にする**計算です。
標準偏差が1なので、2乗した分散も1になります。
$$
\begin{gather}
入力データ\

x=

\left[

\begin{array}{}

  x_{11}&x_{12}&\cdots&x_{1j}\\

  x_{21}&x_{22}&\cdots&x_{2j}\\

  \vdots&\vdots&\ddots&\vdots\\

  x_{i1}&x_{i2}&\cdots&x_{ij}

\end{array}

\right], \ 
x_j = \left[

\begin{array}{}

x_{1j}\\

x_{2j}\\

\vdots\\

x_{ij}

\end{array}

\right] \ \ \\ \\
標準化 \
x_j'=\frac{x_j-\mu_j}{\sigma_j}\ \ \ (\mu_jはj列目の平均値,\ \sigma_jはj列目の標準偏差)
\end{gather}
$$
計算内容は$x_j- \mu_j$の平行移動と$/\sigma_j$のスケーリングのみなので、**データの相対的な構造は変わらない**という特徴があります。
最小値と最大値に指定がない場合や、外れ値が存在する場合に有効です。
##### もう一つの手法
もう一つのスケーリングの手法に、**正規化**があり、データを0~1の間に収めます。
機械学習でデータの前処理を行う場合は、一般的に標準化の方がよく使われます。
### 過学習の回避-L2正則化
モデルを学習する時、パラメータが訓練データに過剰適合する**過学習**に陥り、**訓練データの予測性能に比べて未知データの予測性能がかなり低くなる**ことがあります。
過学習を緩和するには、**モデルの出力がパラメータに依存しすぎないようにする必要**があります。
損失関数に**正則化項=ペナルティ項**を加え、ペナルティ項まで含めた値を損失と考えるようにします。
損失を最小化するという考え方の元、**損失関数に加え、ペナルティ項の最小化を目指すようにモデルが学習する**ようになります。
$$
\begin{gather}
重みベクトル\\
w=\left[
\begin{array}{}
w_1 & w_2 & \cdots & w_j
\end{array}
\right]\\\\
L2正則化項(\lambda は任意の値)\\
\frac{\lambda}{n}||w||^2_2=\frac{\lambda}{n}\sum_jw_j^2\\ \\
損失関数にL2正則化を追加\\
L(w,b)\leftarrow L(w,b)+\frac{\lambda}{n}||w||^2_2\\\\
偏微分した場合\\
\nabla_wL(w,b)\leftarrow\nabla_wL(w,b)+\frac{2\lambda}{n}w\\
\nabla_bL(w,b)\leftarrow\nabla_bL(w,b)+0
\end{gather}
$$
L2正則化を用いて学習した場合、**パラメータは極端に0に近づきます**。
$\lambda$を大きくするほど正則化項の影響が大きくなり、パラメータは0に近い値になりやすくなります。
正則化は入力データに基づいて学習される重みを用いる関係上、**特徴量のスケールが違うと計算が上手く行きません。**
正則化をする場合は、特徴量のスケールを合わせられる**標準化をしておくと安心**です。
##### 他の正則化
L2正則化と同じくらい有名な手法として、**L1正則化**があります。
L2正則化を用いて算出された重みは0に極端に近いことが多いですが、L1正則化の場合は**一部の重みが0になります**。
これには、**スパース化=特徴量選択**という効果があり、出力値の予測対する影響が少ない特徴量をなくすことができます。
(数学的な証明もあるようですが、ここでは飛ばします。)
$$
\begin{gather}
L1正則化\\
||w||_1=\sum_j|w_j|
\end{gather}
$$
また、0でない重みの数を用いる**L0正則化**もあります。
計算量がとても多くなりやすいためあまり使われることはなく、**微分が不可能**なため、機械学習ではほとんど使うことが出来ません。
ディープラーニングでは、正則化の近似として間接的に使われることが多いようです。

```python
# -*- coding : utf-8 -*-
import numpy as np

def Standarize(X):
    """標準化

    Args:
        X (ベクトル): 入力データ
        
    Returns:
        X (ベクトル): 標準化後の入力データ
    """
    return (X - np.mean(X)) / np.std(X)

def L2Regularization(w, a):
    """L2正則化<br>
    実用を考えるならば、モデルの中に組み込む必要がある.<br>
    今回は、関数として書いてみただけ.
    
    Args:
        X (ベクトル): 重みベクトル
        a (float): 正則化係数
        
    Returns:
        w (ベクトル): 正規化後の重みベクトル
    """
    return (a / len(w)) * sum(w ** 2)
```
